```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries loaded")
```

```python
df= pd.read_csv("data/ai4i2020.csv")
df.head()
```

```python
print(df.shape)
```

```python
df.info()
```

```python
print(df["Machine failure"] .value_counts())
print(df["Machine failure"] .value_counts(normalize=True) *100)
```

```python
plt.figure(figsize=(6, 4))
sns.countplot(x="Machine failure", data=df)
plt.title("Machine Failure Distribution")
plt.xlabel("Machine failure (0 = OK, 1 = Failed)")
plt.ylabel("Count")
plt.show()
```

```python
df.describe()
```

```python
features = [
    "Air temperature [K]",
    "Process temperature [K]",
    "Rotational speed [rpm]",
    "Torque [Nm]",
    "Tool wear [min]",
]

print(df.groupby("Machine failure")[features].mean())
```

```python
plt.figure(figsize=(6, 4))
sns.boxplot(x="Machine failure" , y="Torque [Nm]" , data=df)
plt.title("Torque by Machine Failure")
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.heatmap(df[features + ["Machine failure"]].corr(), annot=True, fmt=".2f", cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

```python
df["temp_diff"] = df["Process temperature [K]"] - df["Air temperature [K]"]
df["power"] = df["Torque [Nm]"] * df["Rotational speed [rpm]"] * 2 * np.pi / 60

print(df[["temp_diff", "power", "Machine failure"]].corr()["Machine failure"])
```

```python
plt.figure(figsize=(8, 4))
sns.histplot(data=df, x="power", hue="Machine failure", bins=50, stat="density", common_norm=False)
plt.title("Power by Machine Failure")
plt.show()
```
